```python
from ultralytics import YOLO
import torch

model = YOLO("yolo11m.pt")

results = model.train(
  data='./People-Detection-11/data.yaml',
  epochs=10,
  imgsz=640,
  batch=16,
  cache="disk",
  device='cuda:0',
  )

metrics = model.val()

path = model.export()
```
